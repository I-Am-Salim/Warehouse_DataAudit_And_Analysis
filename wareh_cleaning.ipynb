## Damage Report

**Category**
- Capitalised values

**Quantity**
- Numbers in names: 'two hundred'
- Missing/Null Values

**Price**
- Missing/Null Values

**Last Restocked**
- Missing dates

**Product ID**
- Multiple difference products have the same product ID

**Mismacting Dtypes**
- Quantity should be an integer not string
- Last restocked should be in datetime format not string
- Product ID doesn't need to be integer

**Status**
- Many products have an out of stock status but the values in their quantity columns are not zero

**Making The Columns into Snakecase**

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.strip()
df.columns = df.columns.str.replace(" ","_")
df.columns
```

**Product ID Flag**
- The Product ID is not a valid key/index as many id's map to multiple products at once
- 238 Product IDs map to more than one product
- Can't be used as a key to join tables or a way of counting number of distinct products
- It has been retained so as to not delete any evidence but do not use as the unit of analysis, use each row instead 

```python
products = df.groupby('product_id')['product_name'].nunique()
products
```

**Product Name**
- Padded whitespaces and lowercase
- Must normalise by removing padding and making into a title

```python
df['product_name']=df['product_name'].str.strip()
df['product_name']=df['product_name'].str.title()
df['product_name'].nunique()
```

**Category**
- All uppercase letters must be normalised to title case

```python
df['category'] = df['category'].str.title()
df['category'].unique()
```

**Status**
- 3 values, all in appropriate format
- No cleaning needed

```python
df['status'].value_counts()
```

**Price**
- 207 Missing values out 1000 total values
- There are 4 price values for each product name and the name of the product doesn't indicate which price it is as all product names have the same 4 price values
    - Same goes for category
- No imputation of price as we have no real way of guessing a products price thus we leave as blank

**Quantity**
- Column in string format instead of integer
- The words 'two hundred' appear instead of 200
- Thus two hundred changed to 200 and column is converted to integer

```python
df.groupby("category")['quantity'].unique()
```

```python
df.groupby("product_name")['quantity'].unique()
```

```python
df['quantity'] = df['quantity'].str.replace("two hundred", "200")
df['quantity'] = pd.to_numeric(df['quantity'], errors='coerce').astype('Int64')
df['quantity']
```

**Last Restocked**
- Missing date values left as null
- Converted from string format to US datetime format

```python
df['last_restocked'] = pd.to_datetime(df['last_restocked'], format='%d/%m/%Y')
df['last_restocked'].dtype
```

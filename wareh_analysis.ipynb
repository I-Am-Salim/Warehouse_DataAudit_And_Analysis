**Univariate EDA Conclusions**
1. Every categorical is near-uniform. Category, warehouse, location, supplier, and status all split almost evenly (e.g. status 340/332/328; warehouses 349/332/319). Real inventory is lumpy — bestsellers, dead stock, favored suppliers — so this flatness across every dimension suggests the data was randomly generated rather than drawn from real operations.

2. The "numeric" columns are really a handful of tiers. Quantity takes only 5 distinct values (50, 100, 150, 200, 300) and price only 4 (9.99, 19.99, 29.99, 49.99).

3. A fifth of the key data is missing, and stock status contradicts stock on hand. Quantity, price, and last_restocked are each ~16–21% missing (158 / 207 / 200 rows). Meanwhile there is no row with quantity 0 anywhere, yet 332 rows are marked "Out of Stock" — so the status field disagrees with the quantity field and can't be taken at face value.

**Q3 — Do suppliers differ in how recently they restock?**
- Supplier C restocked the longest average of days ago
- The average number of days since each restocked is fairly even and uniform

```python
latest_date = df['last_restocked'].max()
df['days_ago'] = (latest_date - df['last_restocked']).dt.days
df
```

```python
sup_stock = df.groupby('supplier')['days_ago'].agg(['mean'])
sup_stock['mean'] = sup_stock['mean'].round()
sup_stock
```

**Q8 — How concentrated is inventory value? Top line items and their share of the total.**
- The top 10.0% of rows hold 28% of the total capital.
- The top 20.0% of rows hold 47% of the total capital.
- The top 30.0% of rows hold 60% of the total capital.
- The top 50.0% of rows hold 78% of the total capital.

```python
s = df['inventory_value'].dropna().sort_values(ascending=False).reset_index(drop=True)
for pct in [0.1,0.2,0.3,0.5]:
    n = int(len(s)*pct)
    top_share = s.head(n).sum()/s.sum()
    print(f"The top {pct*100}% of rows hold {top_share:.0%} of the total capital.")
```

**Histogram of Inventory Value of Products**

```python
plt.figure(figsize=(12,8))
sns.set_theme(style='whitegrid', context='talk', palette='colorblind')
ax=sns.histplot(data=df, x='inventory_value',bins=10, kde=True, kde_kws={'bw_adjust':0.7}, edgecolor='white', color='#4C72B0', linewidth=1.2)
ax.xaxis.set_minor_locator(mticker.AutoMinorLocator())
ax.xaxis.set_minor_formatter(mticker.NullFormatter())
ax.yaxis.set_minor_locator(mticker.AutoMinorLocator())
ax.yaxis.set_minor_formatter(mticker.NullFormatter())
ax.grid(which='minor', axis='x', alpha=0.25, linewidth=0.6)
ax.grid(which='minor', axis='y', alpha=0.25,linewidth=0.6)
ax.grid(True, linestyle='--', alpha=0.4, color='gray')
plt.title('Inventory Value of Products', fontweight='bold', fontsize=26, pad=16)
plt.xlabel("Value", labelpad=12);plt.ylabel('No. Of Products', labelpad=12)
ax.axvline(df['inventory_value'].mean(), color='crimson', linestyle='-', linewidth=2, label='Mean')
ax.axvline(df['inventory_value'].median(), color='black', linestyle=':', linewidth=2, label='Median')
ax.legend()
plt.tight_layout()
sns.despine()
plt.savefig('figures/invent_hist', dpi=150, bbox_inches='tight')
plt.show()
plt.close()
```
